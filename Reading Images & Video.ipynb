```python
# Reading videos

capture = cv.VideoCapture('FreeCodeCampTutorial\opencv-course\Resources\Videos\dog.mp4')
```

```python
while True:
    isTrue, frame = capture.read()
    cv.imshow('Video', frame)

    # Stop video from playing indefinitely
    if cv.waitKey(20) & 0xFF==ord('d'):
        break

# Release the capture pointer
capture.release()
cv.destroyAllWindows()
```
